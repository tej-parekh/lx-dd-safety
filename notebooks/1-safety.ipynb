# Safety

Before you touch a drone, a soldering iron, or fly outdoors, you need to understand the risks
involved and how to manage them. This learning experience walks you through the safety analysis
you must complete before starting the *Introduction to Robotics with Drones* course.

Safety is everyone's responsibility.  Please speak up if you observe any unsafe situation during any of the work in this course (or any other time!)

## Intended learning outcomes

After completing this learning experience, learners will be able to:

1. Analyze the safety considerations for operating a robot (drone) using the OSHA framework for
   industrial robot and robot system safety.
2. Explain the FAA rules and regulations that govern operating an unmanned aircraft system (UAS),
   including how to complete the TRUST certification.
3. Identify the institutional (Brown University) and personal (at-home) policies and precautions
   that apply to flying a drone outside of the lab.
4. Set up a soldering station and a flight area that meet the safety requirements for this course.



## 1. OSHA Safety Analysis

Robots — including drones — can injure the people operating and working around them if they are
not used correctly. The Occupational Safety and Health Administration (OSHA) publishes a Technical
Manual on Industrial Robots and Robot System Safety that lays out the standard framework for
thinking about robot safety.

**Read:** the [OSHA Technical Manual, Section IV, Chapter 4 — *Industrial Robots and Robot System Safety*](https://www.osha.gov/otm/section-4-safety-hazards/chapter-4).

**Do:** Identify the **three** hazard categories from the OSHA framework that you think are most
relevant to the drone you will fly in this course. For each one, briefly describe (a) why it
applies to your drone/lab setup and (b) one concrete mitigation you could put in place.

Don't worry if you don't yet know every detail of your drone or lab space — give your best
current understanding; you'll refine this as the course goes on.

1. Impact, Collision, or Other "Struck-by/Caught-between" Hazards:
   This is a relevant risk since we will eventually be flying our drone. Since these drones are made from scratch, there is probably an especially high chance of one acting unexpectedly and striking a student or course staff member. One risk mitigation strategy we should use in response is wearing proper protective equipment and flying the drone in controlled, predictable spaces.

2. Slipping, Tripping, and Falling Hazards:
   The space we will be building the drones in is likely full of laboratory equipment that we could potentially trip on. In addition, many students may be using the space at the same time, making it more cramped. A useful mitigation strategy would be keeping tools and cables organized and making sure walkways remain clear.

3. Electrical Hazards:
   Our drones will use batteries, wires, and other electrical components. Damaged or incorrectly connected components could create electricalchazards. A good mitigation strategy would be disconnecting the battery when modifying the drone's electronics and checking wiring and connections before powering the system on.

## 2. FAA Rules

Because a drone is an unmanned aircraft, flying one is regulated by the Federal Aviation
Administration (FAA), not just by OSHA-style workplace safety rules.

**Do:**

1. Review the FAA's [Unmanned Aircraft Systems](https://www.faa.gov/uas) website.
2. Complete **The Recreational UAS Safety Test (TRUST)** through any FAA-approved test
   administrator. This is free and takes about 30 minutes. Keep the certificate and show it as part of your checkoff.
3. Answer the following four questions:
   1. What are the procedures for flying outdoors, beyond the CIT (or wherever your indoor lab
      space is)?
   2. What is the nearest airport to campus, including heliports?
   3. What are the risks and potential injuries involved in flying a drone?
   4. What are the FAA's requirements for reporting an accident?


1. Before flying outdoors, I would check the location in B4UFLY and make sure there are not current flight restrictions. For an educational flight at a university, the FAA allows educational operations to follow recreational flyer rules. This means completing TRUST, keeping the drone within visual line of sight, yielding to other aircraft, and staying at or below 400 ft in uncontrolled airspace. If the location is in controlled airspace, I would need FAA authorization through LAANC or DroneZone before flying.

2. The nearest active heliport to the CIT is the Rhode Island Hospital Heliport, 1.8 miles by foot from the CIT.

3. A drone can collide with people, buildings, vehicles, or other aircraft. Its propellers can cause cuts or eye injuries. A falling or fast-moving drone can cause blunt force injuries. Batteries and electrical components can also create burn and fire hazards if damaged.

4. An accident must be reported to the FAA within 10 days if it causes a serious injury, loss of consciousness, or more than $500 of damage to property other than the drone itself.

## 4. Soldering Station Safety 

Soldering is hot work and requires precautions for soldering safely.  Watch the soldering tutorial video here: https://youtu.be/Qps9woUGkvI

1.  What safety precautions are missing (or at least, not clearly visible) in this video? 

1. There is no mention of washing your hands after soldering. Solder often contains lead, which can be toxic, so it is important to wash your hands thoroughly after soldering. Also, the rosin fumes produced during soldering can be harmful, so using proper ventilation, such as a fume extractor or hood, is important.
